# Data Preparation

### Selecting revevant subsets of the data

The school information data set contains many columns. We only need the year, school identity, location, and school type information.

Keep the columns SCHOOL_YEAR, NCESSCH, LSTATE, LZIP, SCH_TYPE_TEXT, LEVEL, CHARTER_TEXT

```python
school_information.columns
```

```python
school_information = school_information[
    ['SCHOOL_YEAR', 'NCESSCH', 'LSTATE', 'LZIP', 'SCH_TYPE_TEXT', 'LEVEL', 'CHARTER_TEXT']
]
```

```python
school_information.head()
```

### Renaming columns

```python
edgap.columns
```

```python
edgap = edgap.rename(
    columns = {
       'NCESSCH School ID': 'id', 
       'CT Unemployment Rate': 'rate_unemployment',
       'CT Pct Adults with College Degree': 'percent_college',
       'CT Pct Childre In Married Couple Family': 'percent_married', 
       'CT Median Household Income': 'median_income',
       'School ACT average (or equivalent if SAT score)': 'average_act',
       'School Pct Free and Reduced Lunch': 'percent_lunch',
    }
)
```

```python
edgap.head()
```

```python
school_information.columns
```

```python
school_information = school_information.rename(
    columns = {
       'NCESSCH': 'id', 
       'SCHOOL_YEAR': 'year',
       'LSTATE': 'state',
       'LZIP': 'zip_code', 
       'SCH_TYPE_TEXT': 'school_type',
       'LEVEL': 'school_level',
       'CHARTER_TEXT': 'charter',
    }
)
```

```python
school_information.head()
```

The most poignant subset of columns have been selected and renamed for clarity

## Join data frames

```python
school_information['id'] = school_information['id'].astype('object')
```

```python
school_information.info()
```

### Joining the data frames and call the result df.

```python
df = edgap.merge(
    school_information,
    how = 'left', # use the left join
    on = 'id'     # use id as the key
)
```

```python
df.head()
```

```python
df.info()
```

### Summary
1. We joined the data sets using the school id as the key.
2. We first cast the id as an object to have the same data type for the key in each data set.
3. We used a left join to keep all EgGap data.

Data sets have been joined using a left join and the school id as the key.

### Droping the rows where the average ACT score is missing.

ACT score is the reponse variable of our interest and the percentage of missing values for ACT score is low.

```python
df = df.dropna(subset = ['average_act'])
```

```python
df.isna().sum().to_frame(name = 'Number of Missing Values')
```

If we drop rows that have NaNs for any socioeconomic variables, then we will negatively affect our analysis. We don't want to drop the entire row just because one socioeconomic variable is missing, because we would also lose the useful information that is available in the other variables. We will impute the missing values.

All out-of-range values are now set to NaN. Only highschools who have an average ACT score are kept in the data set. 

## Data Imputation 

Defining the predictor variables to be rate_unemployment, percent_college, percent_married, median_income, percent_lunch, state, and charter.

```python
predictor_variables = [
    'rate_unemployment', 
    'percent_college', 
    'percent_married', 
    'median_income', 
    'percent_lunch', 
    'state', 
    'charter'
]
```

### Replacing missing values in the columns corresponding to predictor variables in the analysis.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer() #models the variable with missing values as a function of other variables
```

### Fitting the imputer using the numerical predictor variables (this can include dummies for categorical variables)

Define the columns selected for the imputation process

```python
numerical_predictors = df[predictor_variables].select_dtypes(include = 'number').columns.to_list()
print(numerical_predictors)
```

Fit the imputer.

```python
imputer.fit(df.loc[:, numerical_predictors]) # learn the relationships between the columns
```

Imputing the missing values in the training data.

```python
df.loc[:, numerical_predictors] = imputer.transform(df.loc[:, numerical_predictors])
```

### Checking for missing values

```python
df.isna().sum().to_frame(name = 'Number of Missing Values')
```

```python
There are no remaining missing values.
```
